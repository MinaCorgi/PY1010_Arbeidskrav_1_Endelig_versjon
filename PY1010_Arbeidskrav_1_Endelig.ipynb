Program som beregner og presenterer årlig totalkostnad for elbil og bensinbil
Jeg forutsetter en årlig kjørelengde på 16 000 km i året, 
og har valgt selvforklarende variabelnavn for å begrense antall Markdown-celler.

Variabler og beregninger

```python
kjørelengde_totalt = 16000.0  # Min forutsetning for valgt kjørelengde
forsikring_el = 5000.00
forsikring_bensin = 7500.00
årsavgift_begge_per_dag = 8.38
dager_i_ett_år = 365.0
årsavgift = int(årsavgift_begge_per_dag * dager_i_ett_år)
bomavgift_el = 0.10
bomavgift_årlig_el = bomavgift_el * kjørelengde_totalt
bomavgift_bensin = 0.30
bomavgift_årlig_bensin = bomavgift_bensin * kjørelengde_totalt
drivstoff_bensin = 1.00
drivstoff_per_år_bensin = drivstoff_bensin * kjørelengde_totalt
drivstoff_el = 0.20
strømpris = 2.00
forbruk_strøm_per_år = kjørelengde_totalt * drivstoff_el
drivstoff_årlig_kostnad_el = strømpris * forbruk_strøm_per_år
total_kostnad_el = forsikring_el + årsavgift + drivstoff_årlig_kostnad_el + bomavgift_årlig_el  # årlig kostndad elbil
total_kostnad_bensin = forsikring_bensin + årsavgift + drivstoff_per_år_bensin + bomavgift_årlig_bensin  # årlig kostnad bensinbil
årlig_kostnadsdifferanse = total_kostnad_bensin - total_kostnad_el  # Viser differansen på kostnadene mellom biltypene
```

Utskrift skjerm for begge biltyper

```python
print('')
print('Utgangspunktet er en årlig kjørelengde på 16 000 km for begge biltyper.')
print('')
print('')
print('Oversikt over de årlige kostnadene for el-bil:')
print('Forsikring: = ',forsikring_el)
print('Trafikkforsikringsavgift: = ',årsavgift)
print('Drivstoffbruk: = ',drivstoff_årlig_kostnad_el)
print('Bomavgift: = ',bomavgift_årlig_el)
print('')
print('Total kostnad per år for elbil = ', total_kostnad_el)
print('')
print('')
print('Oversikt over de årlige kostnadene for bensin-bil:')
print('Forsikring: = ',forsikring_bensin)
print('Trafikkforsikringsavgift: = ',årsavgift)
print('Drivstoffbruk: = ',drivstoff_per_år_bensin)
print('Bomavgift: = ',bomavgift_årlig_bensin)
print('')
print('Total kostnad per år for bensinbil = ', total_kostnad_bensin)
print('')
print('')
print('Den totale kostnadsforskjellen mellom bensin og elbil blir = ', årlig_kostnadsdifferanse)
print('Bensinbil er med disse kostnadsverdiene kr', årlig_kostnadsdifferanse, 'dyrere per år enn en elbil.')
```
